```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Load the image
image_path = '/content/IMG 2.jfif'
original_image = cv2.imread(image_path)

# Check if image was loaded successfully
if original_image is None:
    print(f"Error: Could not load image from {image_path}")
else:
    # Convert the image to grayscale
    gray_image = cv2.cvtColor(original_image, cv2.COLOR_BGR2GRAY)
    print("Image converted to grayscale successfully.")
```

```python
if original_image is not None:
    # Display the original and grayscale images
    plt.figure(figsize=(10, 5))

    plt.subplot(1, 2, 1)
    plt.title('Original Image')
    plt.imshow(cv2.cvtColor(original_image, cv2.COLOR_BGR2RGB)) # Convert BGR to RGB for matplotlib
    plt.axis('off')

    plt.subplot(1, 2, 2)
    plt.title('Grayscale Image')
    plt.imshow(gray_image, cmap='gray')
    plt.axis('off')

    plt.show()

    # Save the grayscale image
    output_path = '/content/IMG_2_grayscale.png'
    cv2.imwrite(output_path, gray_image)
    print(f"Grayscale image saved to {output_path}")
```

```python
def add_salt_and_pepper_noise(image, salt_vs_pepper, amount):
    '''Adds Salt and Pepper noise to an image.'''
    # Create a copy of the image to avoid modifying the original
    noisy_image = np.copy(image)

    # Get image dimensions
    row, col = noisy_image.shape

    # Add Salt noise
    num_salt = np.ceil(amount * noisy_image.size * salt_vs_pepper)
    coords = [np.random.randint(0, i - 1, int(num_salt)) for i in noisy_image.shape]
    noisy_image[coords[0], coords[1]] = 255

    # Add Pepper noise
    num_pepper = np.ceil(amount * noisy_image.size * (1. - salt_vs_pepper))
    coords = [np.random.randint(0, i - 1, int(num_pepper)) for i in noisy_image.shape]
    noisy_image[coords[0], coords[1]] = 0

    return noisy_image

# Define parameters for salt and pepper noise
salt_vs_pepper_ratio = 0.5 # Ratio of salt vs. pepper noise. 0.5 means equal amounts.
amount_of_noise = 0.01 # The proportion of pixels to be affected by noise

# Apply salt and pepper noise to the grayscale image
if 'gray_image' in locals() and gray_image is not None:
    noisy_gray_image = add_salt_and_pepper_noise(gray_image, salt_vs_pepper_ratio, amount_of_noise)
    print("Salt and Pepper noise applied to grayscale image.")
else:
    print("Error: 'gray_image' not found. Please ensure the grayscale conversion cell was executed successfully.")
```

```python
if 'noisy_gray_image' in locals() and noisy_gray_image is not None:
    # Display the noisy grayscale image
    plt.figure(figsize=(8, 6))
    plt.title('Grayscale Image with Salt and Pepper Noise')
    plt.imshow(noisy_gray_image, cmap='gray')
    plt.axis('off')
    plt.show()

    # Optionally save the noisy image
    output_noisy_path = '/content/IMG_2_grayscale_noisy.png'
    cv2.imwrite(output_noisy_path, noisy_gray_image)
    print(f"Noisy grayscale image saved to {output_noisy_path}")
```

```python
# Apply median filter to reduce salt and pepper noise
# The kernel size must be odd, e.g., 3, 5, 7
k_size = 5 # Median filter kernel size

if 'noisy_gray_image' in locals() and noisy_gray_image is not None:
    denoised_image = cv2.medianBlur(noisy_gray_image, k_size)
    print(f"Median filter with kernel size {k_size} applied to the noisy grayscale image.")
else:
    print("Error: 'noisy_gray_image' not found. Please ensure the noise application cell was executed successfully.")
```

```python
if 'denoised_image' in locals() and denoised_image is not None:
    # Display the denoised grayscale image
    plt.figure(figsize=(8, 6))
    plt.title(f'Grayscale Image after Median Filter (k={k_size})')
    plt.imshow(denoised_image, cmap='gray')
    plt.axis('off')
    plt.show()

    # Optionally save the denoised image
    output_denoised_path = '/content/IMG_2_grayscale_denoised.png'
    cv2.imwrite(output_denoised_path, denoised_image)
    print(f"Denoised grayscale image saved to {output_denoised_path}")
```
